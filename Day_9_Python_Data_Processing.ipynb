## Employee Records Data Processing

This notebook demonstrates basic data processing using Python lists,
dictionaries, functions, loops, and summary statistics.

```python
employees = [
    {
        "id": 101,
        "name": "Arun",
        "department": "ECE",
        "salary": 35000,
        "experience": 1
    },
    {
        "id": 102,
        "name": "Priya",
        "department": "CSE",
        "salary": 42000,
        "experience": 2
    },
    {
        "id": 103,
        "name": "Karthik",
        "department": "ECE",
        "salary": 38000,
        "experience": 2
    },
    {
        "id": 104,
        "name": "Divya",
        "department": "IT",
        "salary": 45000,
        "experience": 3
    },
    {
        "id": 105,
        "name": "Rahul",
        "department": "ECE",
        "salary": 50000,
        "experience": 4
    },
    {
        "id": 106,
        "name": "Meena",
        "department": "CSE",
        "salary": 48000,
        "experience": 3
    },
    {
        "id": 107,
        "name": "Vijay",
        "department": "IT",
        "salary": 40000,
        "experience": 2
    },
    {
        "id": 108,
        "name": "Anjali",
        "department": "ECE",
        "salary": 55000,
        "experience": 5
    }
]

print("Dataset created successfully!")
```

```python
print("Employee Records")
print("----------------")

for employee in employees:
    print(employee)
```

```python
total_employees = len(employees)

print("Total number of employees:", total_employees)
```

```python
total_salary = 0

for employee in employees:
    total_salary += employee["salary"]

average_salary = total_salary / len(employees)

print("Average Salary:", average_salary)
```

```python
salaries = []

for employee in employees:
    salaries.append(employee["salary"])

minimum_salary = min(salaries)
maximum_salary = max(salaries)

print("Minimum Salary:", minimum_salary)
print("Maximum Salary:", maximum_salary)
```

```python
total_experience = 0

for employee in employees:
    total_experience += employee["experience"]

average_experience = total_experience / len(employees)

print("Average Experience:", round(average_experience, 2), "years")
```

```python
department_count = {}

for employee in employees:
    department = employee["department"]

    if department in department_count:
        department_count[department] += 1
    else:
        department_count[department] = 1

print("Employees by Department:")
print(department_count)
```

```python
def calculate_average_salary(records):
    total = 0

    for employee in records:
        total += employee["salary"]

    return total / len(records)
```

```python
average = calculate_average_salary(employees)

print("Average Salary:", average)
```

```python
above_average = []

for employee in employees:
    if employee["salary"] > average_salary:
        above_average.append(employee)

print("Employees earning above average salary:")
print()

for employee in above_average:
    print(employee["name"], "-", employee["salary"])
```

```python
def get_employees_by_department(records, department):
    result = []

    for employee in records:
        if employee["department"] == department:
            result.append(employee)

    return result
```

```python
ece_employees = get_employees_by_department(employees, "ECE")

print("ECE Employees:")
print()

for employee in ece_employees:
    print(employee)
```

## Summary Statistics

# Interview Questions

### 1. What is data processing?

Data processing is the process of collecting, organizing,
transforming, and analyzing raw data to obtain useful information.

### 2. Why are functions useful in data processing?

Functions allow us to organize reusable code into separate
blocks. They reduce code repetition and make programs easier
to read, test, maintain, and modify.

### 3. What is structured data?

Structured data is data organized in a predefined format,
usually using rows and columns or clearly defined fields.
Examples include employee records, student records, and
sales records.
